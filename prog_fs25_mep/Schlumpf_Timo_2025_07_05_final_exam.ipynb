# 4 Erstellen eines Dictionaries aus zwei Listen

Gegeben sind zwei Listen: eine Liste von Schlüsseln und eine Liste von Werten. Erstelle ein Dictionary, das Paare von Schlüsseln und Werten enthält. Wenn ein Wert jedoch ein Vielfaches von 3 ist, soll stattdessen "MultiplesOfThree" als Wert eingetragen werden. Gehe davon aus, dass beide Listen die gleiche Länge haben.

> Lösen Sie dies einmal ohne und einmal nur mit **dict-comprehension**.

*Beispiel-Input:*

````python
keys = ["a", "b", "c", "d", "e"]
values = [1, 6, 3, 8, 10]
````

*Beispiel-Output:*

````python
result = {'a': 1, 'b': 'MultiplesOfThree', 'c': 'MultiplesOfThree', 'd': 8, 'e': 10}
````

### 4.1 Ansatz ohne *dict-comprehension*

<div style="text-align: right"> <font color="magenta"> [ ___ / 1 ] </div>

<font color='magenta'>_**Code**_</font>

````python
# your code
out = dict()
for i in range(0,len(keys)):
  if values[i] % 3 == 0:
    out[keys[i]] = 'MultiplesOfThree'
  else:
    out[keys[i]] = values[i]
out    
````

<font color='magenta'>_**Code**_</font>

## 5.1 Vorverarbeitung

<div style="text-align: right"> <font color="magenta"> [ ___  / 4 ] <font color="black"> </div>
<div style="text-align: left"> <font color='purple'>
.
</font> </div>

Um den Text vorzuverarbeiten soll er mit Hilfe der Funktion `preprocess_text()` in Kleinbuchstaben gewandelt, von allen Satzzeichen befreit und in einzelne Wörter aufgeteilt werden. Die Funktion gibt für jede Zeichenketten eine Liste von Wörtern zurück. Stellen Sie sicher, dass diese Liste jedes Wort nur einmal enthält.

> Kleinbuchstaben sind bspw. im `string`-Modul gegeben: `string.ascii_lowercase`.

<font color='magenta'>_**Antwort**_</font>

````python
# your solution
import string
def preprocess_text(txt):
  out = []
  for row in txt:
    row = row.lower()
    for c in row:
      if c in string.punctuation:
        row = row.replace(c, '')
    out.append(row.split(' '))
  return out

preprocess_text(sentences)
  
    
````

Kreditkartennummern bestehen aus einer Reihe von Ziffern, die mit einer Prüfziffer endet. Das ermöglicht Tippfehler bei der Eingabe schnell zu erkennen und so die Korrektheit der Kreditkartennummer zu überprüfen. Die Prüfziffer ist immer die letzte Ziffer der Kreditkartennummer.

Die Prüfsumme kann aus der Kreditkartennummer wie folgt berechnet werden:

+ Die Kreditkartennummer wird aufgespalten: in die reine Kreditkartennummer und separat
die letzte Zahl als Prüfziffer.

+ Beginnend mit der letzten Ziffer, also vom rechten Ende der reinen Kreditkartennummer wird jede zweite Ziffer verdoppelt, also mit 2 multipliziert.
+ Ergibt sich bei der Verdopplung als Ergebnis ein ein zweistelliger Wert, dann werden werden beide Ziffern einfach miteinander addiert.
+ Abschliessend werden alle Werte in einer Summe $s$ zusammengefasst.
  
+ Aus der Summe $s$ ergibt sich die Prüfziffer wie folgt: $10 – ( s~\textrm{modulo}~10 )$.

*Beispiel:*

Wir nehmen an die Kreditkartennummer sei 79927398713. 
Also ist die Prüfziffer 3, da es die letzte Ziffer der Kreditkartennummer ist.
Die reine Kreditkartennummer ist 7992739871, dass ist die zu verarbeitende Zahl.

+ Die Summe $s$ ergibt sich aus dem oben beschriebenen Vorgehen zu 67.

+ Die Prüfziffer ergibt sich aus der Summe $s$ wie folgt: 10 – ( *67* mod 10 ) = 3

+ Da die Prüfziffer 3 der letzten Ziffer der Kreditkartennummer entspricht ist die Kreditkartennummer korrekt.

*Beispiele von korrekten und inkorrekten Kreditkartennummern für allfällige Tests.*

## Aufgabe 8 – NumPy: Array-Berechnungen (10 Punkte)

Gegeben sei ein NumPy-Array mit zufälligen Zahlen. Schreiben Sie eine Funktion, die folgende Schritte ausführt:

Hinweis: 
Ein Z-Wert gibt an, wie viele Standardabweichungen ein Datenpunkt vom Mittelwert entfernt ist.  
Er wird berechnet durch:

$z = \frac{x - \mu}{\sigma}$

Dabei ist $x$ der Wert, $\mu$ der Mittelwert und $\sigma$ die Standardabweichung.  
Z-Werte standardisieren Daten und machen sie vergleichbar.

````python
import numpy as np

def analyse_array(arr):
    # TODO: Mittelwert berechnen
    mean = np.mean(arr)

    # TODO: Standardabweichung berechnen
    std = np.std(arr)

    # TODO: Array normalisieren (Z-Werte)
    z_scores = [((x - mean) / std) for x in arr]

    return z_scores

# Testen mit Zufallsarray
array = np.random.normal(10, 2, 100)
analyse_array(array)
````

## Aufgabe 10 – Objektorientierte Programmierung (10 Punkte)

Implementieren Sie eine Klasse Rechteck, die zwei Attribute breite und höhe hat. Die Klasse soll:

1. Ein __init__-Konstruktor haben (2 Punkte)

2. Eine Methode flaeche() zur Berechnung der Fläche implementieren (2 Punkte)

3. Eine Methode ist_quadrat() zurückgeben, ob es sich um ein Quadrat handelt (2 Punkte)

4. Die __str__() Methode überschreiben, um ein lesbares Format zu geben, z. B. "Rechteck(5 x 10)" (2 Punkte)

5. Ein Klassenattribut anzahl_objekte, das zählt, wie viele Rechtecke erzeugt wurden (2 Punkte)

````python
class Rechteck:
  breite = 0
  hoehe = 0
  anzahl_objekte = 0
  def __init__(self, breite, hoehe):
    self.breite = breite
    self.hoehe = hoehe
    Rechteck.anzahl_objekte += 1

  def flaeche(self):
    return self.breite * self.hoehe

  def ist_quadrat(self):
    return self.breite == self.hoehe

  def __str__(self):
    return f"Rechteck({self.breite} x {self.hoehe})"
````

## Aufgabe 12 – API-Zugriff mit `requests` (10 Punkte)
Verwenden Sie die API `https://api.agify.io`, um das geschätzte Alter für einen Namen abzurufen.

1. Funktion `get_age(name)` schreiben (3 Punkte)
2. `requests.get` mit Parametern verwenden (3 Punkte)
3. Alter aus JSON extrahieren (2 Punkte)
4. Funktion auf mehrere Namen anwenden (2 Punkte)

![Screenshot of api.agify.io](./img/prog_api.png)

````python
import requests

def get_age(name):
    url = "https://api.agify.io"
    params = {
        "name": name
    }
    response = requests.get(url=url,params=params)
    data = response.json()
    age = data['age']
    return age

# Beispiele
names = ["Anna", "Liam", "Mia"]
for name in names:
    print(name, ":", get_age(name))
````

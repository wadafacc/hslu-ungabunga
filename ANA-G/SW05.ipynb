```python
import numpy as np

def f(x):
  return (-x ** 3 + 5 * x - 3) / x

# Nullstellen; Bisektionsmethode
def bisect(f,a,b,e):
  fa = f(a)
  fb = f(b)
  while b-a >= e:
    m = (a + b) / 2
    fm = f(m)
    if fa * fm < 0:
      b = m
      fb = fm
    else:
      a = m
      fa = fm
  print(a,b)

bisect(f,0.5, 1,1E-3)
bisect(f,1, 2,1E-3)
bisect(f,-3, -2,1E-3)
```

```python
import numpy as np

def g(x):
  return (-x ** 3 + 5 * x - 3) / x

def dg(x):
  return -3 * x ** 2 + 5


# Nullstellen; Newtonmethode
def newton(f,df,x0,e):
  x = x0
  dx = f(x) / df(x)
  while np.abs(dx) >= e:
    x = x - dx
    dx = f(x) / df(x)
  print(x)

newton(g,dg,1,1E-15)
newton(g,dg,3,1E-15)
newton(g,dg,-3,1E-15)
```

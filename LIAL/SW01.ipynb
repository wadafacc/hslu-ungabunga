```python
import numpy as np
import matplotlib.pyplot as pl
import numpy.linalg as li
import ipywidgets as widget
```

```python
def Vec(x,y):
  return np.array([
    [x],
    [y]
  ])

def Plot(p, u, c):
    pl.arrow (p [0] [0], p [1] [0], u [0] [0], u [1] [0],
        color = c,
        head_width = 1,
        length_includes_head = True,
        fill = False)


def Scale(vec, s):
  return vec * s

def Add(v, v2):
  return v + v2

def Sub(v, v2):
  return v - v2
```

```python
a = Vec(1,2)
b = Vec(2,4)
print (a + b)
```
